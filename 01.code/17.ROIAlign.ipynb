# ROIalign

```python
import tensorflow as tf
import matplotlib.pyplot as plt
```

# 读取图像

```python
img1 = plt.imread("Trump.jpg")/255.
```

```python
plt.imshow(img1)
```

```python
img2 = plt.imread('Trump2.jpg')/255.
```

```python
plt.imshow(img2)
```

# 构建batch数据

```python
img1 = tf.image.resize(img1,(500,500))
```

```python
img1.shape
```

```python
img1 = tf.convert_to_tensor(img1,dtype=tf.float32)
```

```python
# 增加batch维度
img1 = tf.expand_dims(img1,axis=0)
```

```python
img1.shape
```

```python
img2 = tf.image.resize(img2,(500,500))
```

```python
img2.shape
```

```python
img2 = tf.convert_to_tensor(img2,dtype=tf.float32)
```

```python
# 增加batch维度
img2 = tf.expand_dims(img2,axis=0)
```

```python
img2.shape
```

```python
img = tf.concat([img1,img2],axis=0)
```

```python
img.shape
```

# RoiAlign

```python
roi = tf.image.crop_and_resize(img,[[0.5,0.5,1,1],[0.5,0.5,1,1]],[0,1],crop_size=(100,100))
```

```python
roi.shape
```
